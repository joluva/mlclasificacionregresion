## Importación de Librerias

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

```python
# Definimos la ruta del archivo
file_path = 'AmesHousing.csv'

# Carga de datos
datos = pd.read_csv(file_path)

print(f"Dataset cargado: {datos.shape[0]} filas y {datos.shape[1]} columnas.")
```

#### Para el PCA, selecciono las variables numéricas (que representan las características físicas y de calidad de las viviendas) y excluyo identificadores técnicos como 'Order' y 'PID', además de la variable objetivo 'SalePrice'

```python
# Seleccionamos columnas numéricas
# Excluimos SalePrice (objetivo) e IDs que no aportan valor semántico al PCA
columns_to_exclude = ['SalePrice', 'Order', 'PID']
X = datos.select_dtypes(include=[np.number]).drop(columns=columns_to_exclude, errors='ignore')

# Imputación de valores faltantes con la media
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)

print(f"Columnas seleccionadas para PCA: {X.columns.tolist()}")

```

#### PCA es sensible a las escalas. Las variables deberian tener media 0 y varianza 1 antes de aplicar el modelo.

```python
# Inicializamos el escalador
scaler = StandardScaler()

# Ajustamos y transformamos los datos
X_scaled = scaler.fit_transform(X_imputed)

print("Datos estandarizados correctamente.")
```

#### Aplico el modelo configurado con 16 componentes principales

```python
# Definimos el modelo PCA con n_components=16
n_components = 16
pca = PCA(n_components=n_components)

# Ajusto el modelo y transformamo los datos al nuevo espacio dimensional
X_pca = pca.fit_transform(X_scaled)

# Creamos un DataFrame con los resultados
pca_columns = [f'PC{i+1}' for i in range(n_components)]
datos_pca = pd.DataFrame(data=X_pca, columns=pca_columns)

print(f"Análisis PCA completado. Nueva forma de los datos: {datos_pca.shape}")
datos_pca.head()
```

#### Verificao cuánta información del dataset original logramos retener con estos 16 componentes.

```python
# Varianza explicada por cada componente
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.sum(explained_variance)

print("Varianza explicada por cada componente:")
for i, var in enumerate(explained_variance):
    print(f"PC{i+1}: {var:.4f}")

print(f"\nVarianza total retenida por los 16 componentes: {cumulative_variance:.2%}")
```
